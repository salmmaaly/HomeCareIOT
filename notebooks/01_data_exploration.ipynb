```python
import pandas as pd
```

```python
file_path = "../aruba.txt"

data = []

with open(file_path, "r") as file:
    for line in file:
        parts = line.strip().split()

        date = parts[0]
        time = parts[1]
        sensor = parts[2]
        value = " ".join(parts[3:])

        data.append([date, time, sensor, value])

df = pd.DataFrame(
    data,
    columns=["date", "time", "sensor", "value"]
)

df.head(20)
```

```python
df["timestamp"] = pd.to_datetime(
    df["date"] + " " + df["time"],
    format="mixed"
)

df.head()
```
