```python
df["timestamp"] = pd.to_datetime(df["timestamp"])

df["hour"] = df["timestamp"].dt.hour

def get_period(hour):
    if 5 <= hour < 12:
        return "morning"
    elif 12 <= hour < 17:
        return "afternoon"
    elif 17 <= hour < 21:
        return "evening"
    else:
        return "night"

df["time_period"] = df["hour"].apply(get_period)
```

```python
# Create activity_label
def get_activity(row):
    sensor = row["sensor_type"]
    value = row["value"]

    if sensor == "M":
        if value == 1:
            return "movement"
        else:
            return "no_movement"

    elif sensor == "D":
        if value == 1:
            return "door_open"
        else:
            return "door_closed"


    elif sensor == "T":
        return "temperature"


    else:
        return "unknown"


df["activity_label"] = df.apply(get_activity, axis=1)
```

```python
def extract_activity(value):
    value = str(value).strip()

    if " begin" in value:
        activity = value.replace("ON ", "").replace("OFF ", "").replace(" begin", "")
        return activity

    elif " end" in value:
        activity = value.replace("ON ", "").replace("OFF ", "").replace(" end", "")
        return activity

    return None


def extract_state(value):
    value = str(value).strip()

    if " begin" in value:
        return "begin"

    elif " end" in value:
        return "end"

    return None


df["activity_label"] = df["value"].apply(extract_activity)
df["activity_state"] = df["value"].apply(extract_state)
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
```

```python
activity_df = df[df["activity_label"].notna()].copy()
```

```python
activity_df = activity_df.sort_values("timestamp")
```

```python
activities = []

for activity, group in activity_df.groupby("activity_label"):

    group = group.sort_values("timestamp")

    start_time = None

    for _, row in group.iterrows():

        if row["activity_state"] == "begin":
            start_time = row["timestamp"]

        elif row["activity_state"] == "end" and start_time is not None:

            end_time = row["timestamp"]

            duration = end_time - start_time

            activities.append({
                "activity": activity,
                "start": start_time,
                "end": end_time,
                "duration": duration
            })

            start_time = None
```

```python
activity_duration_df = pd.DataFrame(activities)
```

```python
print(activity_duration_df.head(30))
```

```python
activity_duration_df["duration_minutes"] = (
    activity_duration_df["duration"].dt.total_seconds() / 60
)
```

```python
sleeping = activity_duration_df[
    activity_duration_df["activity"] == "Sleeping"
].copy()

print(sleeping.head(20))
```

```python
print("Average sleep:",
      sleeping["duration_minutes"].mean(), "minutes")

print("Minimum sleep:",
      sleeping["duration_minutes"].min(), "minutes")

print("Maximum sleep:",
      sleeping["duration_minutes"].max(), "minutes")
```

```python
toilet = activity_duration_df[
    activity_duration_df["activity"] == "Bed_to_Toilet"
].copy()

print("Average Bed-to-Toilet:",
      toilet["duration_minutes"].mean(), "minutes")

print("Minimum Bed-to-Toilet:",
      toilet["duration_minutes"].min(), "minutes")

print("Maximum Bed-to-Toilet:",
      toilet["duration_minutes"].max(), "minutes")
```

```python
print(
    activity_duration_df[
        ["activity", "start", "end", "duration_minutes"]
    ].to_string(index=False)
)
```

```python
summary = (
    activity_duration_df
    .groupby("activity")["duration_minutes"]
    .agg(
        occurrences="count",
        total_minutes="sum",
        average_minutes="mean",
        minimum_minutes="min",
        maximum_minutes="max"
    )
    .reset_index()
)

print(summary)
```
